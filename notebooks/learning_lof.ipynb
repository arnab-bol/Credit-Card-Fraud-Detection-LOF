Qn.Why does LOF need nearest neighbors before detecting outliers? -->

Ans:-LOF determines whether a data point is an outlier by comparing the local density around that point with the densities around its neighboring points.

To calculate this local density, the algorithm must first identify the k-nearest neighbors of the point.

Without neighbors:

It cannot measure local density.
It cannot compare densities.
It cannot compute the LOF score.

Therefore, finding the nearest neighbors is the first essential step of the LOF algorithm.


1.k distance:->The k-distance of a point is the distance from that point to its k-th nearest neighbor.(If multiple points are at the same k-distance, include all of them.)

2.Reachability distance :->measures distance while respecting the neighbor's local density.(Anything inside B's neighborhood is treated equally.)(This prevents very dense clusters from dominating the calculations.)

3.Think of LRD as:

Crowdedness Score

*Large LRD = Crowded area
*Small LRD = Sparse area

| LOF Value     |        Meaning                                     |
| ------------------------ | --------------------------------------------------------- |
| ≈ 1                 | Same density as neighbors → Normal                        |
| > 1                  | Lower density than neighbors → Possible outlier           |
| ≫ 1 (e.g., 2, 5, 10) | Much lower density → Strong outlier                       |
| < 1                  | Higher density than neighbors → Definitely not an outlier |

```python
# 1.euclidean distance between A and all other points

def euclidean_distance(point1, point2):

    x1 = point1[0]
    y1 = point1[1]

    x2 = point2[0]
    y2 = point2[1]

    distance = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)

    return distance

A=points[0]
distances=[]

for point in points:
    distance = euclidean_distance(A, point)
    distances.append((point, distance))
    
# print(distances)
distances.sort(key=lambda x: x[1])  #here  we have structure like((1,1), 0) ((2,1), 1.0) so sort using 1st index i.e distance value
for item in distances:
    print(item)
```

```python
# 2.knn

# distances=distances[1:] #remove the first item which is the point itself
# k=2
# neighbors = distances[:k]
# print("Nearest Neighbors:")

# for point, distance in neighbors:
#     print(point, "Distance =", distance)


def get_neighbors(point, points, k):
    distances=[]
    for other_point in points:
        distance = euclidean_distance(point, other_point)
        distances.append((other_point, distance))
        
    distances.sort(key=lambda x: x[1])
    distances=distances[1:] #remove the first item which is the point itself
    neighbors = distances[:k]
    return neighbors

k=2
neighbors = get_neighbors(points[0], points, k)  
for point, distance in neighbors:
    print(point, "Distance =", distance)
```

```python
# 3.k-distance
def get_k_distance(point, points, k):

    neighbors = get_neighbors(point, points, k)

    k_distance = neighbors[-1][1]

    return k_distance

k = 2

for index, point in enumerate(points):

    kd = get_k_distance(point, points, k)

    print(f"Point {index}: {point} -> k-distance = {kd:.3f}")  #kd:3f means 3 decimal points
```

```python
# reachability distance
def reachability_distance(point, neighbor, points, k):

    actual_distance = euclidean_distance(point, neighbor)

    neighbor_k_distance = get_k_distance(neighbor, points, k)

    return max(actual_distance, neighbor_k_distance)

k = 2

neighbors = get_neighbors(points[0], points, k)

for neighbor, distance in neighbors:

    rd = reachability_distance(points[0], neighbor, points, k)

    print(f"Neighbor: {neighbor}, Reachability Distance = {rd:.3f}")
```

```python
# LRD
def local_reachability_density(point, points, k):

    neighbors = get_neighbors(point, points, k)

    reachability_sum = 0

    for neighbor, _ in neighbors:   #here _ is used to ignore the distance value since we only need the neighbor point

        rd = reachability_distance(point, neighbor, points, k)

        reachability_sum += rd

    average_reachability = reachability_sum / k

    lrd = 1 / average_reachability

    return lrd
k = 2

for index, point in enumerate(points):

    lrd = local_reachability_density(point, points, k)

    print(f"Point {index}: {point} -> LRD = {lrd:.3f}")
```

```python
# LOF score
def local_outlier_factor(point, points, k):

    neighbors = get_neighbors(point, points, k)

    neighbor_lrd_sum = 0

    for neighbor, _ in neighbors:

        neighbor_lrd_sum += local_reachability_density(neighbor, points, k)

    average_neighbor_lrd = neighbor_lrd_sum / k

    point_lrd = local_reachability_density(point, points, k)

    lof = average_neighbor_lrd / point_lrd

    return lof

k = 2

for index, point in enumerate(points):

    lof = local_outlier_factor(point, points, k)

    print(f"Point {index}: {point} -> LOF = {lof:.3f}")
```

```python
# The points with the highest LOF scores are the strongest outlier candidates.
lof_scores = []

for point in points:
    score = local_outlier_factor(point, points, k)
    lof_scores.append((point, score))

lof_scores.sort(key=lambda x: x[1], reverse=True)

for point, score in lof_scores:
    print(point, score)
```
